## Part 2 - Engine

### SDR overlap mathematics
A concept is a set of `k = dimensions * sparsity` active bits out of `n = dimensions` (default
`k = 10,000`, `n = 1,000,000`). For two independent random SDRs the overlap is hypergeometric with mean

`E[|A ∩ B|] = k² / n = 100` bits, i.e. a normalised overlap `|A ∩ B| / k = k / n = 1%`.

Unrelated concepts therefore overlap ~1%, while identical concepts overlap 100%.
`bind(a, b)` takes a deterministic random sample of `a ∪ b` of size `k`, so the bound concept keeps about 50% of
each parent (overlap ≈ 0.5) and is nearly orthogonal to everything else. Overlap measures structural relatedness.

### Evolution levels (0-32)
`evolution_level` is a float in `[0, 32]`; see `EVOLUTION_TIERS`. Level 32 is the "integrity PASS" threshold
used by `verify_system_integrity`.

```python
MAX_EVOLUTION_LEVEL = 32.0
EVOLUTION_TIERS = {
    0: "Tabula rasa: empty memory",
    1: "Seeded: first concepts learned",
    4: "Associative: concepts bound into new concepts",
    8: "Persistent: memory survives restarts",
    16: "Structural: multi-level binding chains",
    24: "Verified: claims generated for the formal bridge",
    32: "Optimised: integrity target reached",
}


def tier_for(level: float) -> str:
    """Return the description of the highest tier at or below ``level``."""
    key = max(k for k in EVOLUTION_TIERS if k <= level)
    return EVOLUTION_TIERS[key]


class SuperASI_Core:
    """Sparse-distributed-representation engine with persistent JSON memory.

    Args:
        dimensions: size of the SDR space (default 1,000,000).
        sparsity: fraction of bits active in each SDR (default 1%).
        save_path: JSON file used by ``save_memory`` / ``load_memory``.
        progress: optional callback ``progress(done, total, label)`` for long operations.
    """

    def __init__(self, dimensions: int = 1_000_000, sparsity: float = 0.01,
                 save_path: str = "superasi_memory.json",
                 progress: Optional[Callable[[int, int, str], None]] = None) -> None:
        if dimensions <= 0 or not 0 < sparsity <= 1:
            raise ValueError("dimensions must be > 0 and 0 < sparsity <= 1")
        self._dimensions = dimensions
        self._sparsity = sparsity
        self.active_count = max(1, int(dimensions * sparsity))
        self.save_path = save_path
        self.progress = progress
        self.knowledge_base: Dict[str, str] = {}
        self.concept_sdrs: Dict[str, Set[int]] = {}
        self.evolution_level = 0.0

    @property
    def dimensions(self) -> int:
        return self._dimensions

    @dimensions.setter
    def dimensions(self, value: int) -> None:
        if value != self._dimensions:
            self._dimensions = value
            self.active_count = max(1, int(value * self._sparsity))
            _logger.info("Dimensions updated to %s", value)

    def _sdr_gen(self, seed: str, salt: int = 0) -> Set[int]:
        """Deterministically derive an SDR from ``seed`` (same seed -> same SDR)."""
        digest = hashlib.sha256(f"{seed}_{salt}".encode()).digest()
        rng = random.Random(digest)
        return set(rng.sample(range(self._dimensions), self.active_count))

    def learn(self, concept: str, data: str) -> bool:
        """Store ``data`` under ``concept`` and give it an SDR."""
        self.concept_sdrs[concept] = self._sdr_gen(concept)
        self.knowledge_base[concept] = data
        return True

    def learn_many(self, items: Dict[str, str]) -> int:
        """Bulk ``learn`` with progress reporting. Returns the number learned."""
        total = len(items)
        for i, (concept, data) in enumerate(items.items(), 1):
            self.learn(concept, data)
            if self.progress:
                self.progress(i, total, concept)
        return total

    def bind(self, a: str, b: str, name: str) -> bool:
        """Create concept ``name`` from ``a`` and ``b`` (≈50% overlap with each parent)."""
        sdr_a = self.concept_sdrs.get(a) or self._sdr_gen(a)
        sdr_b = self.concept_sdrs.get(b) or self._sdr_gen(b)
        union = sorted(sdr_a | sdr_b)
        rng = random.Random(hashlib.sha256(f"bind_{a}_{b}_{name}".encode()).digest())
        self.concept_sdrs[name] = set(rng.sample(union, min(self.active_count, len(union))))
        return True

    def recall(self, concept: str) -> str:
        return self.knowledge_base.get(concept, "Concept not found.")

    def overlap(self, a: str, b: str) -> float:
        """Normalised overlap |A ∩ B| / k between two known concepts (0.0 - 1.0)."""
        if a not in self.concept_sdrs or b not in self.concept_sdrs:
            raise KeyError(f"unknown concept: {a if a not in self.concept_sdrs else b}")
        return len(self.concept_sdrs[a] & self.concept_sdrs[b]) / self.active_count

    def compute_overlap_matrix(self) -> Dict[str, Dict[str, float]]:
        names = list(self.concept_sdrs)
        return {a: {b: self.overlap(a, b) for b in names} for a in names}

    def evolve(self, level: float) -> float:
        """Set the evolution level, clamped to [0, 32]. Returns the new level."""
        self.evolution_level = max(0.0, min(MAX_EVOLUTION_LEVEL, float(level)))
        return self.evolution_level

    def save_memory(self) -> None:
        """Atomically write memory to ``save_path`` (temp file + rename)."""
        data = {
            "knowledge_base": self.knowledge_base,
            "concept_sdrs": {k: sorted(v) for k, v in self.concept_sdrs.items()},
            "evolution_level": self.evolution_level,
            "dimensions": self._dimensions,
        }
        tmp = f"{self.save_path}.tmp"
        with open(tmp, "w") as f:
            json.dump(data, f)
        os.replace(tmp, self.save_path)
        _logger.info("Memory saved to %s", self.save_path)

    def load_memory(self) -> bool:
        """Load memory if present. Corrupt files are set aside as ``*.corrupt``."""
        if not os.path.exists(self.save_path):
            return False
        try:
            with open(self.save_path) as f:
                data = json.load(f)
            self.knowledge_base = data.get("knowledge_base", {})
            self.concept_sdrs = {k: set(v) for k, v in data.get("concept_sdrs", {}).items()}
            self.evolution_level = data.get("evolution_level", 0.0)
            self.dimensions = data.get("dimensions", self._dimensions)
        except (OSError, ValueError, TypeError, AttributeError) as exc:
            _logger.warning("Memory file corrupt (%s); starting fresh.", exc)
            os.replace(self.save_path, f"{self.save_path}.corrupt")
            self.knowledge_base, self.concept_sdrs, self.evolution_level = {}, {}, 0.0
            return False
        return True


class FormalVerifier:
    """Bridge to Lean 4. Claims are generated as *text*; checking needs a local ``lean``."""

    @staticmethod
    def _ident(name: str) -> str:
        return "".join(ch if ch.isalnum() or ch == "_" else "_" for ch in name)

    @staticmethod
    def generate_lean_claim(concept_a: str, concept_b: str, result: str) -> str:
        i = FormalVerifier._ident
        return (f"theorem {i(result)}_proof : {i(concept_a)} \u2218 {i(concept_b)} \u2245 {i(result)} :=\n"
                "  sorry -- Proof to be verified by Lean 4 Compiler")

    @staticmethod
    def lean_available() -> bool:
        return shutil.which("lean") is not None

    @staticmethod
    def check(source: str, timeout: int = 60) -> Optional[bool]:
        """Run ``lean`` on ``source``. Returns None if Lean is not installed."""
        if not FormalVerifier.lean_available():
            return None
        with tempfile.TemporaryDirectory() as d:
            path = os.path.join(d, "Claim.lean")
            with open(path, "w") as f:
                f.write(source)
            return subprocess.run(["lean", path], capture_output=True, timeout=timeout).returncode == 0


def bootstrap_brain(brain: SuperASI_Core) -> None:
    """Seed the Omega-Topos pillars."""
    pillars = {
        "CategoryTheory": "The study of mathematical structures and relationships.",
        "ToposTheory": "A generalization of set theory where logic is geometric.",
        "OmegaTopos": "The universal structure where mind and matter are projections of one logic.",
    }
    brain.learn_many(pillars)
    brain.bind("CategoryTheory", "ToposTheory", "StructuralLogic")
    brain.evolve(max(brain.evolution_level, 1.0))


def verify_system_integrity(brain: SuperASI_Core) -> Dict[str, object]:
    """Functional check of memory stores and evolution metrics."""
    return {
        "evolution_level": brain.evolution_level,
        "kb_size": len(brain.knowledge_base),
        "sdr_count": len(brain.concept_sdrs),
        "memory_path": brain.save_path,
        "integrity": "PASS" if brain.evolution_level >= MAX_EVOLUTION_LEVEL else "CHECK_REQUIRED",
    }


def visualize_overlaps(brain: SuperASI_Core) -> None:
    """Heatmap of SDR overlaps (text table if plotting libraries are missing)."""
    concepts = list(brain.concept_sdrs)
    if not concepts:
        return
    matrix = brain.compute_overlap_matrix()
    if not HAS_PLOTTING:
        for a in concepts:
            print(f"{a:32s}", " ".join(f"{matrix[a][b]:.2f}" for b in concepts))
        return
    df = pd.DataFrame([[matrix[a][b] for b in concepts] for a in concepts], index=concepts, columns=concepts)
    plt.figure(figsize=(10, 8))
    sns.heatmap(df, annot=False, cmap="Blues")
    plt.title(f"SDR Overlap Matrix (Level {brain.evolution_level})")
    plt.show()


def build_brain(factory: Callable[..., SuperASI_Core] = SuperASI_Core,
                verifier_factory: Callable[[], FormalVerifier] = FormalVerifier, **kwargs):
    """Dependency-injection entry point: returns ``(brain, verifier)``."""
    return factory(**kwargs), verifier_factory()


def validate_execution_order(*names: str) -> None:
    """Raise a clear error if a required notebook global has not been defined yet."""
    missing = [n for n in names if n not in globals()]
    if missing:
        raise RuntimeError(f"Run earlier cells first; undefined: {', '.join(missing)}")


validate_execution_order("SuperASI_Core", "FormalVerifier", "bootstrap_brain",
                         "verify_system_integrity", "visualize_overlaps")
print("Engine cells defined.")
```

## Part 6 - End-to-end test suite
Each test uses its own small brain and a temporary file, so the real `superasi_memory.json` is never touched.
Covered: learn/recall, determinism, overlap maths, bind, binding chains, persistence round-trip, corrupt-file recovery,
Lean claims, integrity check, evolution clamping and dependency injection.

```python
class SuperASITests(unittest.TestCase):
    def setUp(self):
        self.tmp = tempfile.TemporaryDirectory()
        self.path = os.path.join(self.tmp.name, "mem.json")
        self.brain = SuperASI_Core(dimensions=100_000, save_path=self.path)

    def tearDown(self):
        self.tmp.cleanup()

    def test_learn_recall(self):
        self.brain.learn("A", "alpha")
        self.assertEqual(self.brain.recall("A"), "alpha")
        self.assertEqual(self.brain.recall("missing"), "Concept not found.")

    def test_sdr_deterministic_and_sparse(self):
        self.assertEqual(self.brain._sdr_gen("x"), self.brain._sdr_gen("x"))
        self.assertEqual(len(self.brain._sdr_gen("x")), self.brain.active_count)

    def test_unrelated_overlap_is_low(self):
        self.brain.learn("A", "a"); self.brain.learn("B", "b")
        self.assertLess(self.brain.overlap("A", "B"), 0.05)
        self.assertEqual(self.brain.overlap("A", "A"), 1.0)

    def test_bind_overlaps_both_parents(self):
        self.brain.learn("A", "a"); self.brain.learn("B", "b")
        self.brain.bind("A", "B", "AB")
        self.assertEqual(len(self.brain.concept_sdrs["AB"]), self.brain.active_count)
        for p in ("A", "B"):
            self.assertAlmostEqual(self.brain.overlap("AB", p), 0.5, delta=0.1)

    def test_binding_chain(self):
        for n in "ABC":
            self.brain.learn(n, n)
        self.brain.bind("A", "B", "AB"); self.brain.bind("AB", "C", "ABC")
        self.assertGreater(self.brain.overlap("ABC", "AB"), 0.3)
        self.assertLess(self.brain.overlap("ABC", "A"), self.brain.overlap("ABC", "AB"))

    def test_persistence_roundtrip(self):
        self.brain.learn("A", "a"); self.brain.learn("B", "b"); self.brain.bind("A", "B", "AB")
        self.brain.evolve(8); self.brain.save_memory()
        other = SuperASI_Core(dimensions=100_000, save_path=self.path)
        self.assertTrue(other.load_memory())
        self.assertEqual(other.knowledge_base, self.brain.knowledge_base)
        self.assertEqual(other.concept_sdrs, self.brain.concept_sdrs)
        self.assertEqual(other.evolution_level, 8.0)

    def test_corrupt_memory_recovery(self):
        with open(self.path, "w") as f:
            f.write("{not json")
        self.assertFalse(self.brain.load_memory())
        self.assertTrue(os.path.exists(self.path + ".corrupt"))
        self.assertEqual(self.brain.knowledge_base, {})

    def test_lean_claim(self):
        claim = FormalVerifier.generate_lean_claim("A", "B", "C-D")
        self.assertIn("theorem C_D_proof", claim)
        self.assertIn("sorry", claim)

    def test_integrity_and_evolution_clamp(self):
        self.assertEqual(verify_system_integrity(self.brain)["integrity"], "CHECK_REQUIRED")
        self.assertEqual(self.brain.evolve(99), 32.0)
        self.assertEqual(verify_system_integrity(self.brain)["integrity"], "PASS")
        self.assertEqual(self.brain.evolve(-5), 0.0)

    def test_bootstrap_and_injection(self):
        brain, ver = build_brain(save_path=self.path, dimensions=50_000)
        bootstrap_brain(brain)
        self.assertIn("StructuralLogic", brain.concept_sdrs)
        self.assertIsInstance(ver, FormalVerifier)

    def test_invalid_config(self):
        with self.assertRaises(ValueError):
            SuperASI_Core(dimensions=0)


result = unittest.TextTestRunner(verbosity=1).run(unittest.defaultTestLoader.loadTestsFromTestCase(SuperASITests))
assert result.wasSuccessful(), "Test suite failed"
print("Lean 4 available locally:", FormalVerifier.lean_available())
```

## Part 7 - SDR benchmarks

```python
def bench(label: str, fn: Callable[[], object], repeat: int = 5) -> None:
    times = []
    for _ in range(repeat):
        t = time.perf_counter(); fn(); times.append(time.perf_counter() - t)
    print(f"{label:<34s} best {min(times) * 1000:8.2f} ms   mean {sum(times) / len(times) * 1000:8.2f} ms")


bench_brain = SuperASI_Core(save_path=os.path.join(tempfile.gettempdir(), "superasi_bench.json"))
bench("SDR generation (1M dims, 10k bits)", lambda: bench_brain._sdr_gen("bench"))
bench_brain.learn("A", "a"); bench_brain.learn("B", "b")
bench("bind", lambda: bench_brain.bind("A", "B", "AB"))
bench("overlap", lambda: bench_brain.overlap("A", "B"))
bench("save_memory", bench_brain.save_memory, repeat=3)
bench("load_memory", bench_brain.load_memory, repeat=3)
os.remove(bench_brain.save_path)
```
